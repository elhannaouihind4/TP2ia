```python

from sklearn.feature_extraction.text import CountVectorizer
```

```python

phrases = [
    "Le football est un sport populaire dans le monde",
    "Le cinéma est un art",
    "Le chef prépare un plat délicieux dans le restaurant"
]

print(phrases)
```

```python

vectorizer = CountVectorizer()

matrice = vectorizer.fit_transform(phrases)

print(matrice.toarray())
```

```python
mots = vectorizer.get_feature_names_out()

print(mots)
```

```python

import pandas as pd

df_bow = pd.DataFrame(
    matrice.toarray(),
    columns=mots,
    index=["Phrase 1", "Phrase 2", "Phrase 3"]
)

print(df_bow)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
phrases = [
    "Le football est un sport populaire dans le monde",
    "Le cinéma est un art",
    "Le chef prépare un plat délicieux dans le restaurant"
]
```

```python
tfidf_vectorizer = TfidfVectorizer()

matrice_tfidf = tfidf_vectorizer.fit_transform(phrases)

print(matrice_tfidf.toarray())
```

```python
mots_tfidf = tfidf_vectorizer.get_feature_names_out()

print(mots_tfidf)
```

```python
df_tfidf = pd.DataFrame(
    matrice_tfidf.toarray(),
    columns=mots_tfidf,
    index=["Phrase 1", "Phrase 2", "Phrase 3"]
)
print(df_tfidf.round(3))
```
